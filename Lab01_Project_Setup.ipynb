## Task 1: Repository Directory Structure

````python
import os

folders = ['data/raw', 'data/processed', 'models', 'notebooks', 'src', 'configs', 'logs']

for folder in folders:
    os.makedirs(folder, exist_ok=True)
    open(folder + '/.gitkeep', 'w').close()
````

````python
for root, dirs, files in os.walk('.'):
    dirs[:] = [d for d in dirs if d not in ['.git', '.ipynb_checkpoints', 'aipdd_env', '.venv']]
    level = root.count(os.sep)
    print('    ' * level + os.path.basename(os.path.abspath(root)) + '/')
    for file in files:
        print('    ' * (level + 1) + file)
````
